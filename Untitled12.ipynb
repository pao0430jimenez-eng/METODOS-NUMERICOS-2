**REGLA COMPUESTA DE SIMPSON**


> Métodos Numéricos II







Realizaremos la Regla Compuesta de Simpson para aproximar
una integral definida.



Usaremos el siguiente ejemplo:

La integral es:

$$
\int_0^2 e^{2x}\sin(3x)\,dx
$$

con una tolerancia de:

$$
10^{-4}.
$$

Regla Compuesta de Simpson

La Regla Compuesta de Simpson divide el intervalo $[a,b]$ en $n$
subintervalos, donde $n$ debe ser un número par.

Para calcular a h:

$$
h=\frac{b-a}{n}.
$$

La formula de la Regla Compuesta de Simpson es:

$$
\int_a^b f(x)dx =
\frac{h}{3}
\left[
f(a)+2\sum_{\substack{j=1\\}}^{(\frac{n}{2})-1}f(x_{ij})
+4\sum_{\substack{j=1\\}}^{(\frac{n}{2})}f(x_{ij-1})
+f(b)
\right]- \frac{b-a}{180} h^4 f^{(4)}(\mu)
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```



Para el ej. se tiene:

$$
f(x)=e^{2x}\sin(3x).
$$

```python
def f(x):
    return np.exp(2*x) * np.sin(3*x)
```



Se realiza el algoritmo mediante una función que recibe:

- $f$: función que se desea integrar.
- $a$: límite inferior.
- $b$: límite superior.
- $n$: número de subintervalos.


```python
def simpson_compuesta(f, a, b, n):

    # Simpson se necesita que n sea par
    if n % 2 != 0:
        raise ValueError("El número de subintervalos n debe ser par")

    # Calculamos a h de cada subintervalo
    h = (b - a) / n

    # Generamos los puntos x0, x1, ..., xn
    x = np.linspace(a, b, n + 1)

    # Evaluamos la función en cada punto
    y = f(x)

    # Primer y último término
    suma = y[0] + y[n]

    # Términos que tienen coeficiente 4
    for i in range(1, n, 2):
        suma += 4 * y[i]

    # Términos que tienen coeficiente 2
    for i in range(2, n, 2):
        suma += 2 * y[i]

    # Fórmula de Simpson
    integral = (h / 3) * suma

    return integral
```



Para la integral

$$
\int_0^2 e^{2x}\sin(3x)\,dx
$$

se tienen:

$$
a=0,\qquad b=2 \qquad n=?
$$

La tolerancia solicitada es:

$$
\varepsilon=10^{-4}.
$$

```python
a = 0
b = 2
tol = 1e-4
```

Determinación de $n$ y $h$

Se probarán diferentes valores pares de $n$ hasta encontrar el primero
que produzca un error menor que la tolerancia establecida.

Una vez encontrado $n$, se calcula:

$$
h=\frac{b-a}{n}.
$$

```python
# Valor exacto de la integral
valor_exacto = (
    (np.exp(4) * (2*np.sin(6) - 3*np.cos(6)) + 3)
    / 13
)

print("Valor exacto =", valor_exacto)

# Buscar el primer n par cuyo error sea menor que 10^-4
for n in range(2, 1001, 2):

    aproximacion = simpson_compuesta(f, a, b, n)

    error = abs(valor_exacto - aproximacion)

    if error < tol:
        break

# Calcular h
h = (b - a) / n

print("n =", n)
print("h =", h)
print("Aproximación =", aproximacion)
print("Error =", error)
```

El programa aumenta el número de subintervalos $n$ de dos en dos,
ya que la regla de Simpson requiere que $n$ sea par.

El primer valor que cumple con la tolerancia establecida.

```python
print("=" * 50)
print("       RESULTADO FINAL")
print("=" * 50)
print(f"n = {n}")
print(f"h = {h:.10f}")
print(f"Integral aproximada = {aproximacion:.10f}")
print(f"Error absoluto = {error:.10e}")
print("=" * 50)
```

Gráfica de la función

Se muestra la función utilizada en el ejercicio y los puntos
en los que se aplica la Regla compuesta de Simpson.

```python
x = np.linspace(a, b, 500)
y = f(x)

plt.figure(figsize=(10, 5))

plt.plot(x, y, label=r"$f(x)=e^{2x}\sin(3x)$")

# Puntos utilizados por Simpson
x_simpson = np.linspace(a, b, n + 1)
y_simpson = f(x_simpson)

plt.scatter(
    x_simpson,
    y_simpson,
    s=20,
    label="Puntos de Simpson"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Regla Compuesta de Simpson")
plt.grid(True)
plt.legend()

plt.show()
```
